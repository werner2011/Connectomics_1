```python
neurons = navis.example_neurons(n=3, kind="skeleton") 
print(type(neurons))
print(f"Liczba neuronow {len(neurons)}")
print(neurons.cable_length)
print(neurons.n_nodes)
print(neurons.name)
```

```python
fig, ax = navis.plot2d(
    neurons, 
    method="2d", 
    view=("x", "y")
)
plt.show()
```

```python
neuron = navis.example_neurons(n=1, kind='skeleton') 
print(neuron) 

print("ID", neuron.id) 
print("Nazwa:", neuron.name)
print("Liczba węzłów:", neuron.n_nodes)
print("Cable length:", neuron.cable_length)
print("Soma:", neuron.soma)
print("Root:", neuron.root)
print("Branch points:", neuron.n_branches)
print("Leaf nodes:", neuron.n_leafs)
print("Units:", neuron.units)
```
